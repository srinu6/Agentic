**garak** is "the `nmap` for a language model": you point it at a model and it fires a battery of
**known attack probes**, then tells you which ones landed. It was written by Prof. Leon Derczynski
and is now hosted by NVIDIA.

> Runs in **Google Colab**. You paste your own OpenAI key. We run **one small probe** on
> `gpt-4o-mini` so it finishes in a couple of minutes and costs a few cents.


```python
# install garak
!pip install -q -U garak
```

```python
# paste your key — garak reads it from the OPENAI_API_KEY environment variable
import os
from getpass import getpass
os.environ["OPENAI_API_KEY"] = getpass("Paste your OpenAI API key (sk-...): ")
print("Key set.")
```

## Run one probe

The command below runs a **single** probe — `dan.Dan_11_0`, one well-known "DAN" jailbreak —
against `gpt-4o-mini`. One probe keeps the first run fast and cheap.

- `--model_type openai` → the target is an OpenAI model
- `--model_name gpt-4o-mini` → which model
- `--probes dan.Dan_11_0` → which attack(s) to fire

*(Newer garak also accepts `--target_type` / `--target_name`. If a flag errors, run
`!python -m garak --help` to see what your installed version expects.)*


```python
!python -m garak --model_type openai --model_name gpt-4o-mini --probes dan.Dan_11_0
```

## Read the result

garak prints a line per probe with a **pass rate**. Think of it as the inverse of Attack Success
Rate: a high pass rate means the model resisted most attempts; a low pass rate means the jailbreak
often worked.

garak also writes a full **HTML report** and a `.jsonl` log (the path is printed at the end of the
run — usually under `~/.local/share/garak/` or the current folder). Download the `.html` to see
every prompt and response.


```python
# see where garak saved its report + log files
!ls -t ~/.local/share/garak/garak_runs/ 2>/dev/null | head -5 || ls -t *.jsonl *.html 2>/dev/null | head -5
```

### What you learned
garak is your **broad, automated first pass**: point it at a model, fire known attacks, read the
pass rates and the report. It finds the *known* weaknesses cheaply — freeing human testers to hunt
the novel ones.
