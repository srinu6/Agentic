### The case study in one paragraph
ClaimMitra is a LangChain tool-using agent for SurakshaKavach General Insurance. It answers coverage questions, looks up the logged-in customer's policies and claims, files claims, initiates payouts and updates contact details. It is wrapped in six guardrail layers: an input guard (L1), PII middleware (L2), a tool-result sanitizer (L3), model/tool call limits (L4), Human-in-the-Loop approval for state-changing tools (L5), and an output guard (L6). The legacy core system leaks full KYC (Aadhaar, PAN, bank account) through `get_policy_details`, and one knowledge-base document is poisoned with a hidden instruction to trigger a payout.
